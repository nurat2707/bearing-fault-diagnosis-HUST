```python
import os
import numpy as np
import matplotlib.pyplot as plt
from scipy.io import loadmat

# -----------------------------
# CONFIG
# -----------------------------
root_folder = r"D:\digitransol\HUST dataset"
fs = 51200  # FIXED (correct value)

class_durations = {}


# -----------------------------
# Extract signal
# -----------------------------
def get_signal(mat_data):
    return mat_data["data"].flatten()


# -----------------------------
# MAIN LOOP
# -----------------------------
for class_folder in os.listdir(root_folder):
    class_path = os.path.join(root_folder, class_folder)

    if not os.path.isdir(class_path):
        continue

    print(f"\n===== CLASS: {class_folder} =====")
    total_duration = 0

    for file in os.listdir(class_path):
        if not file.endswith(".mat"):
            continue

        file_path = os.path.join(class_path, file)

        # Load data
        data = loadmat(file_path)
        signal = get_signal(data)

        # -----------------------------
        # Duration
        # -----------------------------
        duration = len(signal) / fs
        total_duration += duration

        print(f"{file} → {duration:.2f} sec")

        # -----------------------------
        # TIME DOMAIN
        # -----------------------------
        time_axis = np.arange(len(signal)) / fs

        # -----------------------------
        # FREQUENCY DOMAIN
        # -----------------------------
        fft_vals = np.fft.fft(signal)
        fft_vals = np.abs(fft_vals)

        freqs = np.fft.fftfreq(len(signal), d=1 / fs)

        # Keep only positive frequencies
        mask = freqs >= 0
        freqs = freqs[mask]
        fft_vals = fft_vals[mask]

        # -----------------------------
        # PLOT
        # -----------------------------
        plt.figure(figsize=(12, 4))

        # Time domain (zoomed)
        plt.subplot(1, 2, 1)
        plt.plot(time_axis[:5000], signal[:5000])
        plt.title(f"{file} - Time Domain")
        plt.xlabel("Time (s)")
        plt.ylabel("Amplitude")

        # Frequency domain (limited range)
        plt.subplot(1, 2, 2)
        plt.plot(freqs[:5000], fft_vals[:5000])
        plt.title(f"{file} - Frequency Domain")
        plt.xlabel("Frequency (Hz)")
        plt.ylabel("Magnitude")

        plt.tight_layout()
        plt.show()

    class_durations[class_folder] = total_duration

# -----------------------------
# FINAL SUMMARY
# -----------------------------
print("\n===== TOTAL DURATION PER CLASS =====")
for cls, dur in class_durations.items():
    print(f"{cls}: {dur:.2f} sec ({dur/60:.2f} min)")
```

```python
import os
import numpy as np
import pandas as pd
from scipy.io import loadmat
from scipy.stats import skew, kurtosis
from scipy.signal import welch

# -------------------------
# CONFIG
# -------------------------
SUBFOLDER_PATH = r"D:\digitransol\HUST dataset\OB"   
OUTPUT_CSV = r"D:\digitransol\features_OB.csv"

fs = 51200
window_size = 2048
overlap = 1024
step = window_size - overlap

# -------------------------
def segment_signal(signal):
    return [
        signal[i:i + window_size]
        for i in range(0, len(signal) - window_size, step)
    ]

# -------------------------
def extract_features(x):

    mean = np.mean(x)
    std = np.std(x)
    rms = np.sqrt(np.mean(x**2))
    skewness = skew(x)
    kurt = kurtosis(x)

    max_val = np.max(x)
    min_val = np.min(x)
    ptp = max_val - min_val

    abs_mean = np.mean(np.abs(x)) + 1e-12
    sqrt_abs_mean = np.mean(np.sqrt(np.abs(x))) + 1e-12

    crest_factor = max_val / (rms + 1e-12)
    shape_factor = rms / abs_mean
    impulse_factor = max_val / abs_mean
    clearance_factor = max_val / (sqrt_abs_mean**2)

    fft_vals = np.abs(np.fft.rfft(x))
    freqs = np.fft.rfftfreq(len(x), 1/fs)

    spectral_mean = np.mean(fft_vals)
    spectral_variance = np.var(fft_vals)
    spectral_skewness = skew(fft_vals)
    spectral_kurtosis = kurtosis(fft_vals)

    prob = fft_vals / (np.sum(fft_vals) + 1e-12)
    spectral_entropy = -np.sum(prob * np.log2(prob + 1e-12))

    peak_freq = freqs[np.argmax(fft_vals)]
    low_band_power = np.sum(fft_vals[freqs < 1000])
    fft_rms = np.sqrt(np.mean(fft_vals**2))

    w_freqs, psd = welch(x, fs=fs)
    welch_peak_freq = w_freqs[np.argmax(psd)]

    psd_prob = psd / (np.sum(psd) + 1e-12)
    welch_entropy = -np.sum(psd_prob * np.log2(psd_prob + 1e-12))

    return [
        mean, std, rms, skewness, kurt,
        max_val, min_val, ptp,
        crest_factor, shape_factor, impulse_factor, clearance_factor,
        spectral_mean, spectral_variance, spectral_skewness, spectral_kurtosis,
        spectral_entropy, peak_freq, low_band_power, fft_rms,
        welch_peak_freq, welch_entropy
    ]

# -------------------------
feature_rows = []

columns = [
    "mean","std","rms","skewness","kurtosis",
    "max","min","ptp",
    "crest_factor","shape_factor","impulse_factor","clearance_factor",
    "spectral_mean","spectral_variance","spectral_skewness","spectral_kurtosis",
    "spectral_entropy","peak_frequency","low_band_power","fft_rms",
    "welch_peak_frequency","welch_entropy",
    "file_name"
]

files = sorted([f for f in os.listdir(SUBFOLDER_PATH) if f.endswith(".mat")])

print("\nFile-wise Feature Extraction:\n")

for file in files:

    filepath = os.path.join(SUBFOLDER_PATH, file)

    try:
        signal = loadmat(filepath)['data'].flatten()
    except:
        print(f"{file}  ---> ERROR reading file")
        continue

    segments = segment_signal(signal)

    # PRINT NUMBER OF ROWS
    print(f"{file}  ---> {len(segments)} segments")

    for seg in segments:
        features = extract_features(seg)
        features.append(file)
        feature_rows.append(features)

# -------------------------
df = pd.DataFrame(feature_rows, columns=columns)
df.to_csv(OUTPUT_CSV, index=False)

print("\nSaved:", OUTPUT_CSV)
print("Total rows:", df.shape[0])
```

```python
import os
import pandas as pd

# -------------------------
# CONFIG
# -------------------------
FOLDER_PATH = r"D:\digitransol\feature"
OUTPUT_FILE = r"D:\digitransol\combined_features.csv"

label_map = {"B": 0, "I": 1, "O": 2, "IB": 3, "IO": 4, "OB": 5, "N": 6}

# -------------------------
all_dfs = []

files = [f for f in os.listdir(FOLDER_PATH) if f.endswith(".csv")]

print("\nProcessing files...\n")

for file in files:

    filepath = os.path.join(FOLDER_PATH, file)

    try:
        df = pd.read_csv(filepath)
    except:
        print(f"{file} → ERROR reading file. Skipping.")
        continue

    if df.empty:
        print(f"{file} → EMPTY file. Skipping.")
        continue

    # -------------------------
    # REMOVE USELESS COLUMN
    # -------------------------
    if "file_name" in df.columns:
        df = df.drop(columns=["file_name"])

    # -------------------------
    # EXTRACT LABEL FROM NAME
    # -------------------------
    # handles: features_IB.csv → IB
    try:
        fault_name = file.split("_")[-1].split(".")[0]
    except:
        print(f"{file} → Cannot extract label. Skipping.")
        continue

    if fault_name not in label_map:
        print(f"{file} → Unknown label ({fault_name}). Skipping.")
        continue

    df["label"] = label_map[fault_name]

    print(f"{file} ---> {df.shape}")

    all_dfs.append(df)

# -------------------------
# MERGE
# -------------------------
if len(all_dfs) == 0:
    print("\nNo valid files found. Check your folder or naming.")
else:
    combined_df = pd.concat(all_dfs, ignore_index=True)

    combined_df.to_csv(OUTPUT_FILE, index=False)

    print("\nFinal Shape:", combined_df.shape)
    print("Saved as:", OUTPUT_FILE)
```

```python
import pandas as pd
import os

# -------------------------
# CONFIG
# -------------------------
FOLDER_PATH = r"D:\digitransol\feature"
OUTPUT_FILE = r"D:\digitransol\combined_features_full.csv"

label_map = {"B": 0, "I": 1, "O": 2, "IB": 3, "IO": 4, "OB": 5, "N": 6}

# -------------------------
# COMBINE
# -------------------------
df_list = []

for file in os.listdir(FOLDER_PATH):
    if file.endswith(".csv"):
        path = os.path.join(FOLDER_PATH, file)
        temp = pd.read_csv(path)

        # Check
        if "file_name" not in temp.columns:
            raise ValueError(f"{file} missing file_name column")

        # Extract fault type from file_name
        temp["fault"] = temp["file_name"].str.extract(r"^([A-Z]+)")

        # Map to label
        temp["label"] = temp["fault"].map(label_map)

        # Remove helper column
        temp.drop(columns=["fault"], inplace=True)

        df_list.append(temp)

# Merge all
df = pd.concat(df_list, ignore_index=True)

# -------------------------
# SAVE
# -------------------------
df.to_csv(OUTPUT_FILE, index=False)

print(" Combined CSV created")
print("Shape:", df.shape)
print("Columns:", df.columns)
```

```python
import pandas as pd

# -------------------------
# LOAD DATA
# -------------------------
FILE_PATH = r"D:\digitransol\combined_features_full.csv"

df = pd.read_csv(FILE_PATH)

print("Original Shape:", df.shape)

# -------------------------
# DROP UNWANTED FEATURES
# -------------------------
drop_features = [
    "mean",
    "skewness",
    "std",
    "fft_rms",
    "shape_factor",
    "clearance_factor",
    "spectral_mean",
    "welch_entropy",
    "welch_peak_frequency",
]

df_clean = df.drop(columns=drop_features, errors="ignore")

print("Cleaned Shape:", df_clean.shape)

# -------------------------
# CHECK RESULT
# -------------------------
print("\nRemaining Columns:\n", df_clean.columns)

print(df_clean.isnull().sum())

```

```python
from sklearn.model_selection import train_test_split

# -------------------------
# STEP 1: UNIQUE FILES (FIX)
# -------------------------
files = df_clean["file_name"].unique().tolist()  # or .to_numpy()

# -------------------------
# STEP 2: ONE LABEL PER FILE
# -------------------------
file_labels = df_clean.groupby("file_name")["label"].first()

# -------------------------
# STEP 3: SPLIT FILES (FIX)
# -------------------------
train_files, test_files = train_test_split(
    files,
    test_size=0.2,
    random_state=42,
    stratify=file_labels.loc[files].to_numpy(),  # force numpy
)
# -------------------------
# STEP 3: SPLIT FILES
# -------------------------
train_files, test_files = train_test_split(
    files, test_size=0.2, random_state=42, stratify=file_labels.loc[files]
)

# -------------------------
# STEP 4: CREATE DATASETS
# -------------------------
train_df = df_clean[df_clean["file_name"].isin(train_files)]
test_df = df_clean[df_clean["file_name"].isin(test_files)]

# -------------------------
# STEP 5: X and y
# -------------------------
X_train = train_df.drop(columns=["label", "file_name"])
y_train = train_df["label"]

X_test = test_df.drop(columns=["label", "file_name"])
y_test = test_df["label"]

# -------------------------
# CHECKS
# -------------------------
print("Train shape:", X_train.shape)
print("Test shape:", X_test.shape)

print("\nTrain class distribution:\n", y_train.value_counts())
print("\nTest class distribution:\n", y_test.value_counts())

# sanity check: no file overlap
print("\nOverlap:", set(train_files) & set(test_files))
```

```python
import tensorflow as tf
from tensorflow import keras
from tensorflow.keras import layers
from sklearn.preprocessing import StandardScaler
from tqdm import tqdm
import numpy as np

# -------------------------
# SCALE DATA
# -------------------------
scaler = StandardScaler()

X_train_scaled = scaler.fit_transform(X_train)
X_test_scaled = scaler.transform(X_test)

# Convert to numpy (safe)
X_train_scaled = np.array(X_train_scaled)
X_test_scaled = np.array(X_test_scaled)
y_train = np.array(y_train)
y_test = np.array(y_test)
```

```python
model = keras.Sequential(
    [
        layers.Dense(128, activation="relu", input_shape=(X_train_scaled.shape[1],)),
        layers.Dense(64, activation="relu"),
        layers.Dense(32, activation="relu"),
        layers.Dense(16, activation="relu"),  # 4th layer
        layers.Dense(7, activation="softmax"),  # 7 classes
    ]
)

model.compile(
    optimizer=keras.optimizers.Adam(learning_rate=0.001),
    loss="sparse_categorical_crossentropy",
    metrics=["accuracy"],
)
```

```python
class TQDMCallback(keras.callbacks.Callback):
    def on_train_begin(self, logs=None):
        self.epochs = self.params["epochs"]
        self.pbar = tqdm(total=self.epochs, desc="Training Neural Network")

    def on_epoch_end(self, epoch, logs=None):
        self.pbar.update(1)
        self.pbar.set_postfix({"loss": logs.get("loss"), "acc": logs.get("accuracy")})

    def on_train_end(self, logs=None):
        self.pbar.close()
```

```python
history = model.fit(
    X_train_scaled,
    y_train,
    epochs=50,
    batch_size=64,
    validation_split=0.1,
    callbacks=[TQDMCallback()],
    verbose=1,
)
```

```python
import seaborn as sns
import matplotlib.pyplot as plt
from sklearn.metrics import confusion_matrix, classification_report

y_pred_probs = model.predict(X_test_scaled)
y_pred = np.argmax(y_pred_probs, axis=1)

print("\nClassification Report:\n")
print(classification_report(y_test, y_pred))

cm = confusion_matrix(y_test, y_pred)

plt.figure(figsize=(8, 6))
sns.heatmap(cm, annot=True, fmt="d", cmap="magma")
plt.title("Confusion Matrix - TensorFlow NN")
plt.xlabel("Predicted")
plt.ylabel("Actual")
plt.show()
```
